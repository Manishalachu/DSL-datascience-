```python
import re
from bs4 import BeautifulSoup
import requests

# Using requests with a proper User-Agent header
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"}
url = "https://en.wikipedia.org/wiki/CNN"

response = requests.get(url, headers=headers)
bs = BeautifulSoup(response.text, "html.parser")

# Find body content container
body_content = bs.find("div", {"id": "bodyContent"})

if body_content:
    links = body_content.find_all("a", href=re.compile("^(/wiki/)((?!:).)*$"))
    print(f"Total wiki links found: {len(links)}\n")

    # Print first 10 links
    for link in links[:10]:
        if "href" in link.attrs:
            print(link.attrs["href"])
else:
    print("Could not find body content container.")
```

```python
import csv
from bs4 import BeautifulSoup
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import requests

url = "https://www.flipkart.com/search?q=mobile+phone"
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36",
    "Accept-Language": "en-US,en;q=0.9",
}

webpage = requests.get(url, headers=headers)

# Print HTTP Status (200 means success)
print("HTTP Status Code:", webpage.status_code)

soup = BeautifulSoup(webpage.content, "html.parser")

# Print the first 500 characters of formatted HTML so Jupyter displays output
prettified_html = soup.prettify()
print("\n--- Output Preview (First 500 chars) ---")
print(prettified_html[:500])
```
